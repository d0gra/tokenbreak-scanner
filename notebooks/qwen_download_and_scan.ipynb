This notebook demonstrates how to use the `tokenbreak-scanner` PyPI package to:
1. Install the scanner from PyPI
2. Download and scan `Qwen/Qwen3-0.6B` in one command
3. Inspect the full vulnerability report programmatically

## Step 2: Scan Qwen3-0.6B via CLI

The scanner will automatically download the required tokenizer artifacts from HuggingFace and produce a vulnerability report.

```python
!tokenbreak-scan Qwen/Qwen3-0.6B --download --trust-remote-code
```

## Step 3: Programmatic Scan via Python SDK

Use the Python API for integration into pipelines, dashboards, or CI gates.

```python
from tokenbreak_scanner.inspector import inspect_model
from tokenbreak_scanner.models import RiskLevel

report = inspect_model("Qwen/Qwen3-0.6B", download=True, trust_remote_code=True)

print("=" * 70)
print("               TOKENBREAK SCANNER REPORT")
print("=" * 70)
print(f"  Model Name:       {report.model_name}")
print(f"  Model Type:       {report.model_type}")
print(f"  Family:           {report.model_family}")
print(f"  Tokenizer Class:  {report.tokenizer_class}")
print(f"  Algorithm:        {report.tokenizer_algorithm.value}")
print(f"  Vocab Size:       {report.vocab_size or 'N/A'}")
print(f"  Confidence:       {report.confidence_score:.2f}")
print(f"  Vulnerable:       {'YES ⚠\uFE0F' if report.vulnerable_to_tokenbreak else 'NO ✅'}")
print(f"  Risk Level:       {report.risk_level.value}")
if report.behavioral_diagnostic:
    bd = report.behavioral_diagnostic
    print(f"  Behavioral:       shifted={bd.shifted}/{bd.total} fragility={bd.fragility:.2f}")
print("=" * 70)
print("  Detection Sources:")
for i, src in enumerate(report.detection_sources, 1):
    print(f"    {i}. [{src.signal}] weight={src.weight:.2f} -> {src.inferred or 'N/A'}")
print("=" * 70)
print("  Recommendation:")
print(f"  {report.recommendation}")
if report.behavioral_diagnostic and report.behavioral_diagnostic.warning:
    print(f"  ⚠️  {report.behavioral_diagnostic.warning}")
print("=" * 70)
```
